# 🚀 Production Deepfake Detector Training
## Fully Automated - Downloads Everything

**Setup (5 min):**
1. Runtime → Change runtime type → **GPU (T4)**
2. Get Kaggle API: https://www.kaggle.com/settings → API → Create Token
3. Upload `kaggle.json` (Files panel, left)
4. **Run All** (Runtime → Run all)

**Then wait 3-4 hours for training to complete!**
